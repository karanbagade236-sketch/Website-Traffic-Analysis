```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("website_traffic.csv")

df["Date"] = pd.to_datetime(df["Date"])

fig = px.line(
    df,
    x="Date",
    y="Number of Website Visitors",
    title="Website Traffic Analysis",
    hover_data=["Number of Website Visitors"]
)

fig.update_traces(
    mode="lines+markers",
    line=dict(color="blue", width=4),
    marker=dict(color="red", size=10)
)

fig.update_xaxes(tickangle=45)

fig.update_layout(
    xaxis_title="Date",
    yaxis_title="Number of Visitors",
    template="plotly_white",
    title_x=0.5
)
fig.write_html("Website.html")
fig.write_html("Website_Traffic_Analysis.png")
fig.show()
```
